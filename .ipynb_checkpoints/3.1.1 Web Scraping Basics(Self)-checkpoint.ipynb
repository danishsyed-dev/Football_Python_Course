```python
import requests
from bs4 import BeautifulSoup
```

```python
url = 'https://www.scrapethissite.com/pages/simple/'
```

```python
response = requests.get(url)
```

```python
response.status_code
```

```python
response.text
```

```python
soup = BeautifulSoup(response.content, "html.parser")
```

```python
title = soup.select_one('h1').text
print(title)
```

```python
soup.select_one('h1')
```

```python
description = soup.select_one('p[class="lead"]').text
print(description)
```

```python
 soup.select_one('p[class="lead"]')
```

```python
description = soup.select_one('p[class="lead"]').text.strip()
print(description)
```

```python
country_names = soup.select('h3[class="country-name"]')
country_names = [x.text.strip() for x in country_names]
print(country_names)
```

```python
#Using Wild card Selector
wild_card_names = soup.select('h3[class*="-name"]')
wild_card_names = [x.text.strip() for x in wild_card_names]
print(wild_card_names)
```

```python
#Using -soup-contain selector
andorra = soup.select_one('h3:-soup-contains("Andorra")').text
print(andorra.strip())
```

```python
population = soup.select('div[class="country-info"] span[class="country-population"]')
print([x.text.strip() for x in population])
```

```python
#using class selector
country_capital = soup.select('span[class="country-capital"]')
print([x.text.strip() for x in country_capital])
```

```python
#using Wild card selector
country_capital = soup.select('span[class*="-capital"]')

country_capital = [x.text.strip()
                   for x in country_capital
                  ]
print(country_capital)
```

```python
#Using zip()
country_with_its_capital = list(zip(country_names, country_capital))
print(country_with_its_capital)
```

```python
#using dict()
country_with_its_capital = dict(zip(country_names, country_capital))

print(country_with_its_capital)
```

```python
#Extracting Country Names using Class selector, then converting to list
country_names = soup.select('h3[class="country-name"]')
country_names = [x.text.strip()
                   for x in country_names
                  ]
#Extracting Country Capitals using Class selector, then converting to list
country_capital = soup.select('span[class="country-capital"]')
country_capital = [x.text.strip()
                   for x in country_capital
                  ]
#Loop and print nicely
for name, capital in zip(country_names, country_capital):
    print(f"{name} → {capital}")
```

```python
#You can make selectors cleaner like this:
country_names = [x.text.strip() for x in soup.select('h3.country-name')]
country_capital = [x.text.strip() for x in soup.select('span.country-capital')]

print(len(country_names), len(country_capital))
```

```python
for name, capital in zip(country_names, country_capital):
    print(f"{name} → {capital}")
```
